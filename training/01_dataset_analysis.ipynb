```python
import pandas as pd
import numpy as np
```

```python
BIAS_TRAIN = "/content/drive/MyDrive/AI-News-Perspective-Analyzer/datasets/article-bias/train-00000-of-00001.parquet"

STANCE_TRAIN = "/content/drive/MyDrive/AI-News-Perspective-Analyzer/datasets/stance/train_merged.csv"
```

```python
bias_df = pd.read_parquet(BIAS_TRAIN)

stance_df = pd.read_csv(STANCE_TRAIN, encoding="latin1")
```

```python
# Remove duplicate rows
bias_df = bias_df.drop_duplicates()
stance_df = stance_df.drop_duplicates()

# Remove extremely short articles
bias_df = bias_df[bias_df["content"].str.len() >= 100]
stance_df = stance_df[stance_df["articleBody"].str.len() >= 100]

# Reset indices
bias_df = bias_df.reset_index(drop=True)
stance_df = stance_df.reset_index(drop=True)
```

```python
bias_df.to_parquet(
    "/content/drive/MyDrive/AI-News-Perspective-Analyzer/datasets/article-bias/train_clean.parquet",
    index=False
)

stance_df.to_csv(
    "/content/drive/MyDrive/AI-News-Perspective-Analyzer/datasets/stance/train_clean.csv",
    index=False
)
```
